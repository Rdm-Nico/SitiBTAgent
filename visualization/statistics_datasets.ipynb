# Statistiche del datasets

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
DATA_FOLDER="../data/training_dataset/train/train_generated.jsonl"
```

```python
df = pd.read_json(path_or_buf=DATA_FOLDER, lines=True)
```

```python
def computeLen(sample):
    return len([msg['content'] for msg in sample if msg['role'] == 'user'])
```

```python
df["len_msg"] = df["messages"].map(computeLen)
```

```python
df.groupby("len_msg").count()["id"]
```

```python
sns.displot(df, x="len_msg", kde=True)
```
